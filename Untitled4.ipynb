```python
# Create training pairs: (image, detected_objects, audio_files)

# We need:
# 1. Your annotated images from Roboflow
# 2. The objects in each image (from labels)
# 3. Matching audio files for those objects

# Download your Roboflow dataset
from roboflow import Roboflow

rf = Roboflow(api_key="UwYqx73NVviMjqK4RvA5")
project = rf.workspace("samo-aawvg").project("elenii_shepherd-p8j8z")
dataset = project.version(3).download("yolov8")

print(f"✅ Roboflow dataset downloaded: {dataset.location}")
```

```python
# Get YOLO class names
yolo_classes = yolo_model.names  # {0: 'ATM machines', 1: 'Appliances', ...}

# Match with audio
matched_pairs = {}
missing_audio = []

for class_id, class_name in yolo_classes.items():
    if class_name in audio_mapping:
        matched_pairs[class_name] = {
            'class_id': class_id,
            'audio_file': audio_mapping[class_name],
            'audio_path': f"{hausa_path}/{audio_mapping[class_name]}"
        }
    else:
        missing_audio.append(class_name)

print(f"\n✅ Matched: {len(matched_pairs)} classes have both vision AND audio")
print(f"⚠️  Missing audio: {len(missing_audio)} classes")
print(f"\nSample matches:")
for i, (obj, data) in enumerate(list(matched_pairs.items())[:5]):
    print(f"  {obj} → {data['audio_file']}")
```

```python
# Reload audio files (since you just added more)
hausa_path = "/content/drive/MyDrive/Elenii_hausa_recording"
hausa_files = os.listdir(hausa_path)

# Create fresh mapping
audio_mapping = {}
for file in hausa_files:
    obj_name = file.split(" ")[0]
    audio_mapping[obj_name] = file

print(f"Updated audio files: {len(audio_mapping)} objects")
print("Sample:", list(audio_mapping.keys())[:10])
```

```python
# Now install and use semantic matching
!pip install sentence-transformers scikit-learn

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load embedding model
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')

# Get embeddings for audio objects
audio_objects = list(audio_mapping.keys())
audio_embeddings = embedding_model.encode(audio_objects)

print(f"✅ Created embeddings for {len(audio_objects)} audio objects")
```

```python
# Smart matching with similarity threshold
matched_pairs_smart = {}
similarity_threshold = 0.5  # 50% similarity minimum

for class_id, class_name in yolo_classes.items():
    # Get embedding for YOLO class name
    class_embedding = embedding_model.encode([class_name])

    # Calculate similarity with all audio objects
    similarities = cosine_similarity(class_embedding, audio_embeddings)[0]

    # Find best match
    best_idx = similarities.argmax()
    best_similarity = similarities[best_idx]
    best_match = audio_objects[best_idx]

    if best_similarity > similarity_threshold:
        matched_pairs_smart[class_name] = {
            'class_id': class_id,
            'matched_audio': best_match,
            'audio_file': audio_mapping[best_match],
            'audio_path': f"{hausa_path}/{audio_mapping[best_match]}",
            'similarity': best_similarity
        }

print(f"\n✅ Smart matching: {len(matched_pairs_smart)} classes matched!")
print(f"⚠️  Unmatched: {len(yolo_classes) - len(matched_pairs_smart)} classes")

# Show some smart matches
print("\nSample smart matches:")
for i, (yolo_name, data) in enumerate(list(matched_pairs_smart.items())[:15]):
    print(f"  '{yolo_name}' → '{data['matched_audio']}' (similarity: {data['similarity']:.2f})")
```

```python
# COMPLETE ELENII SHEPHERD SYSTEM
# Vision Detection + Audio Playback Integration

import IPython.display as ipd
from IPython.display import Image as IPImage, display
import matplotlib.pyplot as plt
from gtts import gTTS
import io

# ============================================
# ELENII SHEPHERD CLASS - Complete System
# ============================================
class EleniiShepherdSystem:
    def __init__(self, yolo_model, audio_mapping, embedding_model, audio_embeddings, audio_objects, audio_path):
        self.yolo_model = yolo_model
        self.audio_mapping = audio_mapping
        self.embedding_model = embedding_model
        self.audio_embeddings = audio_embeddings
        self.audio_objects = audio_objects
        self.audio_path = audio_path
        self.matched_pairs = matched_pairs_smart  # From previous step

    def detect_objects(self, image_path, confidence=0.25):
        """Run YOLO detection on image"""
        results = self.yolo_model(image_path, conf=confidence)
        return results[0]

    def find_audio(self, detected_class, similarity_threshold=0.5):
        """Find matching audio for detected object"""
        # First check if we already matched it
        if detected_class in self.matched_pairs:
            return self.matched_pairs[detected_class]['audio_path'], 'matched'

        # Try semantic similarity
        class_embedding = self.embedding_model.encode([detected_class])
        similarities = cosine_similarity(class_embedding, self.audio_embeddings)[0]

        best_idx = similarities.argmax()
        best_similarity = similarities[best_idx]
        best_match = self.audio_objects[best_idx]

        if best_similarity > similarity_threshold:
            audio_file = self.audio_mapping[best_match]
            return f"{self.audio_path}/{audio_file}", f'similar to {best_match}'

        # Fallback: TTS
        return None, 'no_match'

    def generate_tts_fallback(self, text):
        """Generate speech for objects without audio"""
        try:
            tts = gTTS(text=text, lang='en', slow=False)
            audio_buffer = io.BytesIO()
            tts.write_to_fp(audio_buffer)
            audio_buffer.seek(0)
            return ipd.Audio(audio_buffer.read())
        except:
            return None

    def process_image(self, image_path, play_audio=True, save_result=True):
        """
        Complete Elenii Shepherd pipeline:
        Image → Detection → Audio playback
        """
        print(f"\n{'='*60}")
        print(f"📸 Processing: {image_path}")
        print('='*60)

        # Detect objects
        results = self.detect_objects(image_path)

        # Show annotated image
        if save_result:
            results.save('elenii_result.jpg')
            display(IPImage('elenii_result.jpg'))

        # Process each detection
        if len(results.boxes) > 0:
            print(f"\n🔍 Detected {len(results.boxes)} objects:\n")

            for box in results.boxes:
                class_id = int(box.cls[0])
                confidence = float(box.conf[0])
                class_name = self.yolo_model.names[class_id]

                print(f"  • {class_name} ({confidence*100:.1f}%)")

                # Find and play audio
                if play_audio:
                    audio_path, match_type = self.find_audio(class_name)

                    if audio_path:
                        print(f"    🔊 Playing Hausa audio ({match_type})")
                        display(ipd.Audio(audio_path))
                    else:
                        print(f"    🤖 Generating TTS fallback")
                        tts_audio = self.generate_tts_fallback(class_name)
                        if tts_audio:
                            display(tts_audio)

                print()
        else:
            print("\n❌ No objects detected")

        return results

# ============================================
# INITIALIZE THE COMPLETE SYSTEM
# ============================================
elenii_system = EleniiShepherdSystem(
    yolo_model=yolo_model,
    audio_mapping=audio_mapping,
    embedding_model=embedding_model,
    audio_embeddings=audio_embeddings,
    audio_objects=audio_objects,
    audio_path=hausa_path
)

print("✅ COMPLETE ELENII SHEPHERD SYSTEM READY!")
print(f"   - Vision: {len(yolo_model.names)} object classes")
print(f"   - Audio: {len(audio_mapping)} Hausa recordings")
print(f"   - Matched: {len(matched_pairs_smart)} classes with audio")

# ============================================
# TEST THE COMPLETE SYSTEM
# ============================================
print("\n" + "="*60)
print("TESTING COMPLETE ELENII SHEPHERD SYSTEM")
print("="*60)

# Test on validation images
test_image_path = '/content/Elenii_Shepherd-3/valid/images'
test_images = os.listdir(test_image_path)

# Test on 3 random images
import random
sample_images = random.sample(test_images, 3)

for img_name in sample_images:
    img_path = f"{test_image_path}/{img_name}"
    elenii_system.process_image(img_path)
    print("\n" + "-"*60 + "\n")

# ============================================
# SAVE COMPLETE SYSTEM
# ============================================
print("\n" + "="*60)
print("SAVING COMPLETE SYSTEM")
print("="*60)

# Save system configuration
import pickle

system_config = {
    'model_path': '/content/drive/MyDrive/elenii_models/elenii_trained_model/weights/best.pt',
    'audio_mapping': matched_pairs_smart,
    'audio_path': hausa_path,
    'language': 'hausa',
    'embedding_model': 'all-MiniLM-L6-v2',
    'matched_classes': len(matched_pairs_smart),
    'total_classes': len(yolo_model.names),
    'version': 'v1.0'
}

config_path = '/content/drive/MyDrive/elenii_models/elenii_system_config.pkl'
with open(config_path, 'wb') as f:
    pickle.dump(system_config, f)

print(f"✅ System configuration saved to: {config_path}")
print("\nElenii Shepherd is ready for deployment!")
```

```python
# Real-time webcam detection for Elenii Shepherd
from IPython.display import display, Image as IPImage, clear_output
import cv2
from google.colab.patches import cv2_imshow
import time

def elenii_realtime_webcam(duration_seconds=60, detection_interval=2):
    """
    Real-time Elenii Shepherd detection
    duration_seconds: how long to run
    detection_interval: seconds between detections (to avoid spam)
    """

    print("🚀 STARTING ELENII SHEPHERD REAL-TIME MODE")
    print(f"Running for {duration_seconds} seconds...")
    print(f"Detecting every {detection_interval} seconds")
    print("\nPress Ctrl+C in terminal to stop early\n")

    # Start webcam capture via JavaScript
    from IPython.display import Javascript, display
    from google.colab import output
    from base64 import b64decode
    import io
    from PIL import Image

    # JavaScript for continuous webcam
    js_code = """
    const video = document.createElement('video');
    const canvas = document.createElement('canvas');

    async function startWebcam() {
        const stream = await navigator.mediaDevices.getUserMedia({video: true});
        document.body.appendChild(video);
        video.srcObject = stream;
        video.play();

        return new Promise((resolve) => {
            video.onloadedmetadata = () => {
                canvas.width = video.videoWidth;
                canvas.height = video.videoHeight;
                resolve(video);
            };
        });
    }

    async function captureFrame() {
        canvas.getContext('2d').drawImage(video, 0, 0);
        return canvas.toDataURL('image/jpeg', 0.8);
    }

    startWebcam();
    """

    display(Javascript(js_code))
    time.sleep(5)  # Keep the delay to allow webcam to initialize

    start_time = time.time()
    last_detection = 0
    frame_count = 0

    try:
        while (time.time() - start_time) < duration_seconds:
            current_time = time.time()

            # Only detect at intervals to avoid spam
            if current_time - last_detection >= detection_interval:
                # Capture frame by directly calling the JS function
                from google.colab.output import eval_js
                data = eval_js("captureFrame()") # Directly call the function

                if data:
                    # Save frame
                    binary = b64decode(data.split(',')[1])
                    frame_path = f'frame_{frame_count}.jpg'
                    with open(frame_path, 'wb') as f:
                        f.write(binary)

                    # Run Elenii Shepherd detection
                    clear_output(wait=True)
                    print(f"\n⏱️  Time: {int(current_time - start_time)}s")
                    elenii_system.process_image(frame_path, play_audio=True, save_result=True)

                    last_detection = current_time
                    frame_count += 1

            time.sleep(0.5)

    except KeyboardInterrupt:
        print("\n\n🛑 Stopped by user")

    print(f"\n✅ Elenii Shepherd ran for {int(time.time() - start_time)} seconds")
    print(f"   Processed {frame_count} frames")

# Run real-time detection
print("Starting Elenii Shepherd real-time mode...")
elenii_realtime_webcam(duration_seconds=60, detection_interval=3)
```
